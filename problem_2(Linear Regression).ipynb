```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("student_study_dataset.csv")
```

```python
df.head(5)
```

```python
df.isnull().sum()
```

```python
df['hours_studied'].fillna(df['hours_studied'].mean(),inplace=True)
```

```python
df['exam_score'].fillna(df['exam_score'].mean(),inplace=True)
```

```python
df.head(5)
```

```python
X = df[['hours_studied']]
Y = df['exam_score']
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(X,Y)
```

```python
study_hour = [[3]]
```

```python
YP = model.predict(study_hour)
```

```python
YP
```

```python
plt.scatter(X,Y)
plt.plot(X,model.predict(X))
plt.xlabel("Hour studies")
plt.ylabel("Exam score")
plt.title("Student Score Prediction")
plt.show()
```
