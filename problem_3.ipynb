```python
#Predict car mileage (km/l) from
##engine displacement (cc).
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("car_mileage_dataset.csv")
```

```python
df.head(5)
```

```python
df.isnull().sum()
```

```python
df['engine_cc'].fillna(df['engine_cc'].mean(),inplace=True)
```

```python
df['mileage_kmpl'].fillna(df['mileage_kmpl'].mean(),inplace=True)
```

```python
df.head(5)
```

```python
X = df[['engine_cc']]
Y = df['mileage_kmpl']
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(X,Y)
```

```python
engin = [[1903],[1800]]
```

```python
YP = model.predict(engin)
```

```python
YP
```

```python
plt.scatter(X,Y)
plt.plot(X,model.predict(X),c = 'red')
plt.xlabel("Engin size")
plt.ylabel("mileage KM/hr")
plt.title("Predict car mileage")
plt.show()
```
