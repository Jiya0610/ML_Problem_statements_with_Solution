```python
# Group products in an e-commerce
# store by price, rating, and return
# rate.
```

```python
import pandas as pd
```

```python
df = pd.read_csv("ecommerce_product_kmeans_dataset.csv")
```

```python
df.head()
```

```python
X = df[['price','avg_rating','return_rate']]
```

```python
from sklearn.cluster import KMeans
```

```python
model = KMeans(n_clusters=3)
```

```python
df['cluster'] = model.fit_predict(X)
```
