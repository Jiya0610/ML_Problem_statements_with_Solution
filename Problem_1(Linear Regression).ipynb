```python
#Predict house price based on size
#(sq ft) of the house.
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("house_price_dataset.csv")
```

```python
df.head(5)
```

```python
df.isnull().sum()
```

```python
df['area_sqft'].fillna(df['area_sqft'].mean(),inplace=True)
```

```python
df['price'].fillna(df['price'].mean(),inplace=True)
```

```python
df.head(5)
```

```python
X = df[['area_sqft']]
Y = df['price']
type(X),type(Y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(X,Y)
```

```python
q = [[0],[5],[7]]
```

```python
yp = model.predict(q)
```

```python
yp
```

```python
plt.scatter(X, Y)
plt.plot(X, model.predict(X))
plt.xlabel("Area (sq ft)")
plt.ylabel("Price")
plt.title("House Price Prediction")
plt.show()
```
