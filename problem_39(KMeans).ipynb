```python
# Cluster news articles into topics
# using TF-IDF word frequency
# vectors.
```

```python
import pandas as pd
```

```python
df = pd.read_csv("news_articles_kmeans_dataset.csv")
```

```python
df.head()
```

```python
X = df.drop('article_id',axis=1)
```

```python
from sklearn.cluster import KMeans
```

```python
model = KMeans(n_clusters=3)
```

```python
df['cluster'] = model.fit_predict(X)
```
